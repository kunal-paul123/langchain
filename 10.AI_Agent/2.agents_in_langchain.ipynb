```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
import requests
```

```python
from langchain_community.tools import DuckDuckGoSearchRun

search_tool = DuckDuckGoSearchRun()
```

```python
@tool
def get_weather_data(city: str) -> str:
    """
    thi function fetches the current weather data for a given city
    """
    url = f"http://api.weatherstack.com/current?access_key=41bbfdd67f75b7668f3326d7834eff59&query={city}"

    response = requests.get(url)

    return response.json()
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
from langchain_classic.agents import create_react_agent, AgentExecutor
```

```python
from langsmith import Client

client = Client()
prompt = client.pull_prompt("hwchase17/react")
```

```python
agent = create_react_agent(
    llm=llm,
    tools=[search_tool, get_weather_data],
    prompt=prompt
)
```

```python
agent_executor = AgentExecutor(
    agent=agent,
    tools=[search_tool, get_weather_data],
    verbose=True
)
```

```python
response = agent_executor.invoke({"input": "Find the capital of West Bengal, then find the current weather condition?"})
```
